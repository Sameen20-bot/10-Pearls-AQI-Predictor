```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

from sklearn.model_selection import RandomizedSearchCV, TimeSeriesSplit

from xgboost import XGBRegressor
```

```python
from pathlib import Path

CWD = Path.cwd()
PROJECT_ROOT = CWD.parent.parent
PROCESSED_DIR = PROJECT_ROOT / "Data" / "Processed"

print(PROCESSED_DIR)
```

```python
df = pd.read_csv(PROCESSED_DIR / "Aqi_Data_One_Day_Prediction_New.csv", index_col="time", parse_dates=True)
df.info()
```

```python
df.head()
```

# Train And Test Split

```python
df.columns
```

```python
X = df.drop(columns = ['Day_1_Future_AQI'])
Y = df['Day_1_Future_AQI']
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, shuffle = False)
```

```python
len(X_train)
```

```python
len(X_test)
```

# Building XGBoost Model For Day One

```python
model = XGBRegressor(n_estimators = 520, 
                     learning_rate = 0.033,
                     max_depth = 3, 
                     random_state = 42, 
                     colsample_bytree = 0.7, 
                     subsample = 0.7,
                     min_child_weight = 5,
                     gamma=0.1,
                     reg_alpha=0.1,
                     reg_lambda=1.5,
                    ) 
```

```python
model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)
```

# Evaluating the Model

```python
R2_Score = r2_score(Y_test, y_pred)
Mae = mean_absolute_error(Y_test, y_pred)
Mean_Square_Error = mean_squared_error(Y_test, y_pred)
Root_Mean_Square_Error = np.sqrt(Mean_Square_Error)

print("Evaluation Metrics: ")
print(f"R2 Score: {R2_Score}")
print(f"Mean Absolute Error: {Mae}")
print(f"Mean Square Error: {Mean_Square_Error}")
print(f"Root Mean Square Error: {Root_Mean_Square_Error}")
```

# Hyperparameter Tuning Grid Search

```python
n_estimators     = [int(x) for x in np.linspace(start=100, stop=1000, num=10)]
max_depth        = [1, 2, 3, 4, 5, 6, 7, 8, 10]
learning_rate    = [0.01, 0.05, 0.1, 0.15, 0.2]
subsample        = [0.6, 0.8, 1.0]
colsample_bytree = [0.6, 0.8, 1.0]

random_grid = {'n_estimators': n_estimators,
               'max_depth': max_depth,
               'learning_rate': learning_rate,
               'subsample': subsample,
               'colsample_bytree': colsample_bytree}
```

```python
model_XGB = XGBRegressor(n_jobs = -1, random_state = 42)
```

```python
xgb_random = RandomizedSearchCV(
    estimator = model_XGB,
    param_distributions = random_grid,
    n_iter = 150,
    cv = TimeSeriesSplit(n_splits=5),
    scoring = 'r2',
    verbose = 2,
    random_state = 42,
    n_jobs = 1
)
```

```python
xgb_random.fit(X_train, Y_train)
```

```python
print("Best Params:", xgb_random.best_params_)
print("Best CV R2 :", xgb_random.best_score_)
```

```python
best_random =  xgb_random.best_estimator_
y_pred_random_search = best_random.predict(X_test)
print("R2 Score:", r2_score(Y_test, y_pred_random_search))
```
