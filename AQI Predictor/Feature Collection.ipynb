# Weather Data Collection From Open Meteo

```python
import requests
import pandas as pd
```

```python
url = "https://archive-api.open-meteo.com/v1/archive"

params = {
	"latitude": 24.8608,
	"longitude": 67.0104,
	"start_date": "2023-08-01",
	"end_date": "2026-08-01",
	"hourly": ["temperature_2m", "wind_speed_10m", "relative_humidity_2m", "surface_pressure"],
	"timezone": "auto",
}

response = requests.get(url, params)
```

```python
data = response.json()
```

```python
weatherData = pd.DataFrame(data["hourly"])
```

```python
weatherData.head()
```

```python
weatherData.info()
```

```python
weatherData.time = pd.to_datetime(weatherData.time)
```

```python
weatherData.time = weatherData.time.dt.tz_localize('Asia/Karachi')
```

```python
weatherData.head()
```

```python
weatherData.info()
```

# AQI Data Collection from Open Meteo

```python
url2 = "https://air-quality-api.open-meteo.com/v1/air-quality"

params2 = {
	"latitude": 24.8608,
	"longitude": 67.0104,
	"hourly": ["pm10", "pm2_5", "carbon_monoxide", "nitrogen_dioxide", "sulphur_dioxide", "ozone", "us_aqi"],
	"timezone": "auto",
	"start_date": "2023-08-01",
	"end_date": "2026-08-01",
}

response2 = requests.get(url2, params2)
```

```python
data2 = response2.json()
```

```python
aqiData = pd.DataFrame(data2["hourly"])
```

```python
aqiData.head()
```

```python
aqiData.info()
```

```python
aqiData.time = pd.to_datetime(aqiData.time)
```

```python
aqiData.time = aqiData.time.dt.tz_localize('Asia/Karachi')
```

```python
aqiData.head()
```

```python
aqiData.info()
```

```python
print(aqiData["time"].min())
print(aqiData["time"].max())
print(len(aqiData))
```

# Merge The Weather And AQI Data

```python
AqiData = pd.merge(weatherData, aqiData, on = 'time')
```

```python
AqiData.head()
```

```python
AqiData.info()
```

# Adding New Features

### Time Based Features

```python
AqiData["Hour"] = AqiData["time"].dt.hour
AqiData["Day"] = AqiData["time"].dt.day
AqiData["Month"] = AqiData["time"].dt.month
```

```python
AqiData.head()
```

# AQI Current hour karachi

```python
AqiData.to_csv("AqiDataHourKarachi.csv")
```

### AQI Change Rate

```python
AqiData["AQI Change Rate"] = AqiData["us_aqi"].diff()
```

```python
AqiData.head(100)
```

```python
AqiData["AQI Change Rate"] = AqiData["AQI Change Rate"].fillna(0)
```

```python
AqiData.info()
```

```python
AqiData.to_csv("AqiDataHourChangeRateKarachi.csv")
```

# 1 Day Prediction Aqi

```python
AqiData["1 Day Future Aqi"] = AqiData['us_aqi'].shift(-24)
```

```python
AqiData.info()
```

```python
AqiData_One_Day = AqiData.dropna()
```

```python
AqiData_One_Day.info()
```
