# AQI Category Understanding

```python
day1["us_aqi"].max()
```

```python
day1["us_aqi"].min()
```

```python
labels = ["Good", "Moderate", "Unhealthy for Sensitive", "Unhealthy", "Very Unhealthy", "Hazardous"]
bins = [0, 50, 100, 150, 200, 300, 500]

day1['AQI-Category'] = pd.cut(day1["us_aqi"],bins,labels=labels,include_lowest=True)
```

```python
day1['AQI-Category'].value_counts().sort_index().plot(kind='bar', figsize=(12,4))
plt.ylabel('Hour')
plt.show()
```

```python
day1.head()
```

```python
day1['AQI-Category'].value_counts()
```

```python
day1['AQI-Category'].value_counts()/len(day1["us_aqi"])*100
```

# Note:
1. Moderate(51-100): 76.5% of hours have a moderate AQI level. This is Karachi's typical air quality.
2. Unhealthy for Sensitive(101-150): Unhealthy for Sensitive is about 20.15%, meaning sensitive groups face air quality concerns roughly one-fifth of the time.
3. Unhealthy(151-200): Unhealthy is about 3.11%, but is serious mostly are short length pollutant events.
4. No record for Very Unhealthy and Hazardous.

# Temporal Patterns

```python
plt.figure(figsize=(12,4))
day1.groupby("hour")["us_aqi"].mean().plot(figsize=(12,4), marker="o")
plt.title("AQI MEAN HOURLY OF DAY")
plt.ylabel("Mean AQI")
plt.show()
```

```python
plt.figure(figsize=(12,4))
day1.groupby("dayofweek")["us_aqi"].mean().plot(kind = 'bar')
plt.title("AQI MEAN BY DAY OF WEEK")
plt.ylabel("Mean AQI")
plt.show()
```

```python
plt.figure(figsize=(12,4))
day1.groupby("month")["us_aqi"].mean().plot(kind = 'bar')
plt.title("AQI MEAN BY MONTH")
plt.ylabel("Mean AQI")
plt.show()
```

### Note:
1. As you can see in a mean chart of hourly. The AQI is flat between 87 to 88 from hour 0 to 13 which is (12 of midnight till 2 afternoon). And rises
   from 3 afternoon and 7 of evening is the peak.
2. And for mean day chart, AQI almost same.
3. AQI mean by month, the highest is in January that is above 100, and lowest is in the month of August that is below 80.

# Seasonal Decomposition

```python
result = seasonal_decompose(day1['us_aqi'], model='additive', period=24)
```

```python
result.seasonal[:168].plot(figsize=(12, 6))
plt.show()
```

```python
result.trend.plot(figsize=(12, 6))
plt.show()
```

```python
result.resid.plot(figsize=(12, 6))
plt.show()
```

### Note: There is a visible seasonal pattern cycle. And increasing and decreasing trend, and a residual.

# Auto Correlation Function

```python
fig, ax = plt.subplots(figsize=(16,6))
plot_acf(day1["us_aqi"], lags=24, ax=ax)
ax.set_title("24 Hourly AQI Auto Correlation Function (1 Day ACF)")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(16,6))
plot_acf(day1["us_aqi"], lags=48, ax=ax)
ax.set_title("48 Hourly AQI Auto Correlation Function (2 Day ACF)")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(16,6))
plot_acf(day1["us_aqi"], lags=72, ax=ax)
ax.set_title("72 Hourly AQI Auto Correlation Function (3 Day ACF)")
plt.show()
```

### Note: 
1. Lag 24 has a correlation almost to 0.75.
2. Lag 48 has a correlation almost to 0.58.
3. Lag 72 has a correlation almost to 0.48.
